#### Introduction

Road damage such as potholes and cracks can cause accidents, vehicle
damage, traffic problems, and increased maintenance costs. Traditional
road inspection methods rely heavily on manual monitoring, which can be
time-consuming and difficult to perform frequently across large road
networks.

RoadGuard AI is an AI-based road damage analysis system designed to
analyze road images and identify different types of road damage.

The project focuses specifically on Indian road conditions and uses
image processing and deep learning techniques to analyze road images,
classify damage types, evaluate model performance, and provide an
interpretable assessment of road condition.

#### Problem Definition

Road damage such as potholes and cracks can lead to road accidents,
vehicle damage, traffic problems, and increased maintenance costs.
Traditional road inspection methods rely heavily on manual monitoring,
which can be time-consuming, costly, and difficult to perform frequently
across large road networks.

This project aims to develop a Road Damage Detection and Assessment
System using Image Processing and Deep Learning techniques that
automatically analyzes road images from Indian road conditions.

The system will classify different types of road damage such as
longitudinal cracks, transverse cracks, alligator cracks, and potholes.
The project will further evaluate different deep learning models and
develop a severity assessment methodology to provide a meaningful
road-condition analysis.

#### Dataset Description

The project uses the India subset of the Road Damage Dataset 2022
(RDD2022).

The dataset contains road images collected from Indian road conditions
and corresponding damage annotations.

### Phase 2 — Dataset & Annotation Analysis

```python
CLASS_NAMES = {
    0: "D00 - Longitudinal Crack",
    1: "D10 - Transverse Crack",
    2: "D20 - Alligator Crack",
    3: "D40 - Pothole"
}

CLASS_NAMES
```

#### Analyze all annotations

```python
def analyze_annotations(image_dir, label_dir):

    image_files = sorted([
        p for p in image_dir.iterdir()
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
    ])

    total_images = len(image_files)
    images_with_damage = 0
    images_without_damage = 0
    total_boxes = 0

    class_counts = {class_id: 0 for class_id in CLASS_NAMES}
    boxes_per_image = []

    for image_path in image_files:

        label_path = label_dir / f"{image_path.stem}.txt"

        if not label_path.exists():
            continue

        with open(label_path, "r") as file:
            lines = [
                line.strip()
                for line in file
                if line.strip()
            ]

        # No annotation = no detected damage
        if len(lines) == 0:

            images_without_damage += 1
            boxes_per_image.append(0)

            continue

        images_with_damage += 1
        boxes_per_image.append(len(lines))
        total_boxes += len(lines)

        for line in lines:

            parts = line.split()

            class_id = int(parts[0])

            if class_id not in CLASS_NAMES:
                print("WARNING: Unknown class ID:", class_id)
                continue

            class_counts[class_id] += 1

    return {
        "total_images": total_images,
        "images_with_damage": images_with_damage,
        "images_without_damage": images_without_damage,
        "total_boxes": total_boxes,
        "class_counts": class_counts,
        "boxes_per_image": boxes_per_image
    }
```

#### Run analysis

```python
def analyze_annotations_sample(image_dir, label_dir, sample_size=500):

    image_files = sorted([
        p for p in image_dir.iterdir()
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
    ])

    # Use a sample for quick analysis
    image_files = image_files[:sample_size]

    images_with_damage = 0
    images_without_damage = 0
    total_boxes = 0

    class_counts = {class_id: 0 for class_id in CLASS_NAMES}

    for image_path in image_files:

        label_path = label_dir / f"{image_path.stem}.txt"

        if not label_path.exists():
            continue

        with open(label_path, "r") as file:
            lines = [
                line.strip()
                for line in file
                if line.strip()
            ]

        if not lines:
            images_without_damage += 1
            continue

        images_with_damage += 1
        total_boxes += len(lines)

        for line in lines:
            class_id = int(line.split()[0])

            if class_id in class_counts:
                class_counts[class_id] += 1

    return {
        "images_analyzed": len(image_files),
        "images_with_damage": images_with_damage,
        "images_without_damage": images_without_damage,
        "total_boxes": total_boxes,
        "class_counts": class_counts
    }


quick_analysis = analyze_annotations_sample(
    TRAIN_IMAGES,
    TRAIN_LABELS,
    sample_size=500
)

quick_analysis
```

```python
# ============================================
# Draw YOLO Bounding Boxes
# ============================================

def draw_yolo_annotations(image_path, label_path, class_names):

    image = Image.open(image_path).convert("RGB")
    image = np.array(image)

    height, width = image.shape[:2]

    with open(label_path, "r") as file:
        lines = [
            line.strip()
            for line in file
            if line.strip()
        ]

    annotated_image = image.copy()

    for line in lines:

        parts = line.split()

        class_id = int(parts[0])

        x_center = float(parts[1])
        y_center = float(parts[2])
        box_width = float(parts[3])
        box_height = float(parts[4])

        # Convert normalized coordinates
        x_center *= width
        y_center *= height
        box_width *= width
        box_height *= height

        x1 = int(x_center - box_width / 2)
        y1 = int(y_center - box_height / 2)

        x2 = int(x_center + box_width / 2)
        y2 = int(y_center + box_height / 2)

        # Keep coordinates inside image
        x1 = max(0, x1)
        y1 = max(0, y1)
        x2 = min(width - 1, x2)
        y2 = min(height - 1, y2)

        cv2.rectangle(
            annotated_image,
            (x1, y1),
            (x2, y2),
            (255, 0, 0),
            3
        )

        label = class_names[class_id]

        cv2.putText(
            annotated_image,
            label,
            (x1, max(y1 - 10, 20)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.65,
            (255, 0, 0),
            2
        )

    return annotated_image
```

### Phase 3 — Damage Region Extraction & Image Processing

#### Damage Region Extraction

The original dataset contains complete road images with YOLO bounding-box annotations. Since the objective of the CNN component is to classify road-damage types, the annotated bounding boxes are used to extract the corresponding damage regions from the original images.

This reduces the influence of irrelevant image regions such as sky, buildings, vehicles and surrounding scenery and allows the CNN to focus on the road-damage region.

```python
# ============================================
# Damage Region Extraction
# ============================================

def extract_damage_regions(image_path, label_path):
    
    image = Image.open(image_path).convert("RGB")
    image = np.array(image)

    height, width = image.shape[:2]

    crops = []

    with open(label_path, "r") as file:
        lines = [
            line.strip()
            for line in file
            if line.strip()
        ]

    for line in lines:

        parts = line.split()

        class_id = int(parts[0])

        x_center = float(parts[1])
        y_center = float(parts[2])
        box_width = float(parts[3])
        box_height = float(parts[4])

        # Convert normalized YOLO coordinates
        x_center *= width
        y_center *= height
        box_width *= width
        box_height *= height

        # Convert to corner coordinates
        x1 = int(x_center - box_width / 2)
        y1 = int(y_center - box_height / 2)
        x2 = int(x_center + box_width / 2)
        y2 = int(y_center + box_height / 2)

        # Keep coordinates within image boundaries
        x1 = max(0, x1)
        y1 = max(0, y1)
        x2 = min(width, x2)
        y2 = min(height, y2)

        # Extract crop
        crop = image[y1:y2, x1:x2]

        if crop.size == 0:
            continue

        crops.append({
            "class_id": class_id,
            "class_name": CLASS_NAMES[class_id],
            "crop": crop,
            "bbox": (x1, y1, x2, y2)
        })

    return crops
```

#### Resize the crop

```python
# ============================================
# Image Preprocessing
# ============================================

IMG_SIZE = (224, 224)

def preprocess_damage_crop(crop):
    
    resized = cv2.resize(
        crop,
        IMG_SIZE,
        interpolation=cv2.INTER_AREA
    )

    # Normalize pixel values
    normalized = resized.astype(np.float32) / 255.0

    return normalized
```

### Phase 4 — Build the CNN Dataset

#### CNN Dataset Preparation

The YOLO annotations provide the location and category of each road-damage instance. These annotations are converted into a structured metadata dataset containing the source image, damage class and bounding-box coordinates.

The metadata will be used to dynamically extract and preprocess damage regions during CNN training

```python
# ============================================
# Create CNN Metadata from YOLO Annotations
# ============================================

def create_metadata(image_dir, label_dir, split_name):

    records = []

    image_files = sorted([
        p for p in image_dir.iterdir()
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
    ])

    for image_path in image_files:

        label_path = label_dir / f"{image_path.stem}.txt"

        if not label_path.exists():
            continue

        with open(label_path, "r") as file:
            lines = [
                line.strip()
                for line in file
                if line.strip()
            ]

        # Skip images without damage annotations
        for line in lines:

            parts = line.split()

            class_id = int(parts[0])

            x_center = float(parts[1])
            y_center = float(parts[2])
            box_width = float(parts[3])
            box_height = float(parts[4])

            records.append({
                "split": split_name,
                "image_path": str(image_path),
                "label_path": str(label_path),
                "class_id": class_id,
                "class_name": CLASS_NAMES[class_id],
                "x_center": x_center,
                "y_center": y_center,
                "box_width": box_width,
                "box_height": box_height
            })

    return pd.DataFrame(records)
```

```python
# ============================================
# Create Metadata - First 500 Training Images
# ============================================

def create_metadata_sample(image_dir, label_dir, split_name, sample_size=500):

    records = []

    image_files = sorted([
        p for p in image_dir.iterdir()
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
    ])[:sample_size]

    for image_path in image_files:

        label_path = label_dir / f"{image_path.stem}.txt"

        if not label_path.exists():
            continue

        with open(label_path, "r") as file:
            lines = [
                line.strip()
                for line in file
                if line.strip()
            ]

        for line in lines:

            parts = line.split()

            class_id = int(parts[0])

            records.append({
                "split": split_name,
                "image_path": str(image_path),
                "label_path": str(label_path),
                "class_id": class_id,
                "class_name": CLASS_NAMES[class_id],
                "x_center": float(parts[1]),
                "y_center": float(parts[2]),
                "box_width": float(parts[3]),
                "box_height": float(parts[4])
            })

    return pd.DataFrame(records)


train_metadata_sample = create_metadata_sample(
    TRAIN_IMAGES,
    TRAIN_LABELS,
    "train",
    sample_size=500
)

print("Number of damage regions:", len(train_metadata_sample))

train_metadata_sample.head()
```

#### Extract a crop directly from metadata

```python
# ============================================
# Extract Crop from Metadata
# ============================================

def crop_from_metadata(row):

    image = Image.open(row["image_path"]).convert("RGB")
    image = np.array(image)

    height, width = image.shape[:2]

    x_center = row["x_center"] * width
    y_center = row["y_center"] * height
    box_width = row["box_width"] * width
    box_height = row["box_height"] * height

    x1 = int(x_center - box_width / 2)
    y1 = int(y_center - box_height / 2)
    x2 = int(x_center + box_width / 2)
    y2 = int(y_center + box_height / 2)

    x1 = max(0, x1)
    y1 = max(0, y1)
    x2 = min(width, x2)
    y2 = min(height, y2)

    crop = image[y1:y2, x1:x2]

    return crop
```

####  CNN Data Pipeline

The annotated damage regions are converted into TensorFlow datasets for CNN training and validation. Each sample is dynamically loaded from the original road image using its YOLO bounding-box coordinates.

The extracted damage region is resized to 224 × 224 pixels and normalized to the range [0, 1].

```python
# ============================================
# Load Metadata
# ============================================

train_metadata = pd.read_csv(TRAIN_METADATA_FILE)
val_metadata = pd.read_csv(VAL_METADATA_FILE)

print("Training samples:", len(train_metadata))
print("Validation samples:", len(val_metadata))
```

#### Create numerical class labels

0 → D00
1 → D10
2 → D20
3 → D40

```python
# ============================================
# Numerical Class Labels
# ============================================

CLASS_IDS = sorted(CLASS_NAMES.keys())

CLASS_TO_INDEX = {
    class_id: index
    for index, class_id in enumerate(CLASS_IDS)
}

INDEX_TO_CLASS = {
    index: CLASS_NAMES[class_id]
    for class_id, index in CLASS_TO_INDEX.items()
}

print("CLASS_TO_INDEX:")
print(CLASS_TO_INDEX)

print("\nINDEX_TO_CLASS:")
print(INDEX_TO_CLASS)
```

#### Create the crop loader

```python
# ============================================
# Load and Preprocess Damage Crop
# ============================================

IMG_HEIGHT = 224
IMG_WIDTH = 224
IMG_CHANNELS = 3


def load_damage_crop(row):

    image_path = row["image_path"]

    # Load image
    image = Image.open(image_path).convert("RGB")
    image = np.array(image)

    height, width = image.shape[:2]

    # YOLO normalized coordinates
    x_center = float(row["x_center"]) * width
    y_center = float(row["y_center"]) * height

    box_width = float(row["box_width"]) * width
    box_height = float(row["box_height"]) * height

    # Convert to corner coordinates
    x1 = int(x_center - box_width / 2)
    y1 = int(y_center - box_height / 2)

    x2 = int(x_center + box_width / 2)
    y2 = int(y_center + box_height / 2)

    # Keep coordinates inside image
    x1 = max(0, x1)
    y1 = max(0, y1)
    x2 = min(width, x2)
    y2 = min(height, y2)

    # Crop
    crop = image[y1:y2, x1:x2]

    # Safety check
    if crop.size == 0:
        raise ValueError(f"Empty crop generated for {image_path}")

    # Resize
    crop = cv2.resize(
        crop,
        (IMG_WIDTH, IMG_HEIGHT),
        interpolation=cv2.INTER_AREA
    )

    # Normalize
    crop = crop.astype(np.float32) / 255.0

    return crop
```

#### Create TensorFlow generator

```python
# ============================================
# TensorFlow Data Generator
# ============================================

def data_generator(metadata):

    for _, row in metadata.iterrows():

        image = load_damage_crop(row)

        label = CLASS_TO_INDEX[
            int(row["class_id"])
        ]

        yield image, label
```

#### Create training dataset

```python
# ============================================
# Create Training Dataset
# ============================================

output_signature = (
    tf.TensorSpec(
        shape=(IMG_HEIGHT, IMG_WIDTH, IMG_CHANNELS),
        dtype=tf.float32
    ),
    tf.TensorSpec(
        shape=(),
        dtype=tf.int32
    )
)

train_dataset = tf.data.Dataset.from_generator(
    lambda: data_generator(train_metadata),
    output_signature=output_signature
)

print(train_dataset)
```

#### Create validation dataset

```python
# ============================================
# Create Validation Dataset
# ============================================

val_dataset = tf.data.Dataset.from_generator(
    lambda: data_generator(val_metadata),
    output_signature=output_signature
)

print(val_dataset)
```

#### Batch the data

```python
BATCH_SIZE = 32

train_dataset = (
    train_dataset
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

val_dataset = (
    val_dataset
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)
```

#### Class Weights

```python
# ============================================
# Calculate Class Weights
# ============================================

from sklearn.utils.class_weight import compute_class_weight

y_train = train_metadata["class_id"].values

class_weights_array = compute_class_weight(
    class_weight="balanced",
    classes=np.array(CLASS_IDS),
    y=y_train
)

class_weights = {
    int(class_id): float(weight)
    for class_id, weight in zip(
        CLASS_IDS,
        class_weights_array
    )
}

print("Class weights:")

for class_id, weight in class_weights.items():
    print(
        f"{CLASS_NAMES[class_id]}: {weight:.4f}"
    )
```
