# RoadGuard AI — Final AI Termwork Notebook

**Pipeline:** Dataset → Digital Image Processing → YOLO annotation/localization → CNN classification → VGG16 transfer learning → YOLO object detection → final inference.

Classes: D00 Longitudinal Crack, D10 Transverse Crack, D20 Alligator Crack, D40 Pothole.

```python
from pathlib import Path
import random, shutil, yaml, cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import VGG16
from tensorflow.keras.applications.vgg16 import preprocess_input
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.utils.class_weight import compute_class_weight
from ultralytics import YOLO

SEED=42
random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)

DATASET_DIR=Path('RoadGuard_India')
TRAIN_IMAGES=DATASET_DIR/'train/images'; TRAIN_LABELS=DATASET_DIR/'train/labels'
VAL_IMAGES=DATASET_DIR/'val/images'; VAL_LABELS=DATASET_DIR/'val/labels'
TEST_IMAGES=DATASET_DIR/'test/images'; TEST_LABELS=DATASET_DIR/'test/labels'
OUTPUT_DIR=Path('RoadGuard_AI_Output'); CROP_DIR=OUTPUT_DIR/'cnn_crops'; YOLO_DIR=OUTPUT_DIR/'yolo'

CLASS_NAMES={0:'D00 - Longitudinal Crack',1:'D10 - Transverse Crack',2:'D20 - Alligator Crack',3:'D40 - Pothole'}
CLASS_FOLDERS={0:'D00_Longitudinal',1:'D10_Transverse',2:'D20_Alligator',3:'D40_Pothole'}
IMG_SIZE=224; BATCH_SIZE=32

for p in [TRAIN_IMAGES,TRAIN_LABELS,VAL_IMAGES,VAL_LABELS,TEST_IMAGES,TEST_LABELS]: assert p.exists(), f'Missing: {p.resolve()}'
print('Dataset:', DATASET_DIR.resolve())
```

## 1. Dataset summary

```python
def files(folder, suffix): return list(folder.glob(f'*{suffix}'))
for split in ['train','val','test']:
    print(split, 'images =', len([p for p in (DATASET_DIR/split/'images').iterdir() if p.suffix.lower() in {'.jpg','.jpeg','.png'}]), 'labels =', len(files(DATASET_DIR/split/'labels','.txt')))
print('\nClasses:')
for k,v in CLASS_NAMES.items(): print(k, v)
```

## 2. YOLO annotation statistics

```python
def read_labels(path):
    out=[]
    if not path.exists(): return out
    for line in path.read_text().splitlines():
        a=line.split()
        if len(a)==5: out.append((int(a[0]),*map(float,a[1:])))
    return out

def annotation_stats(label_dir):
    counts={i:0 for i in CLASS_NAMES}; total=0; damaged=0
    for p in label_dir.glob('*.txt'):
        rows=read_labels(p)
        if rows: damaged+=1
        for r in rows:
            if r[0] in counts: counts[r[0]]+=1; total+=1
    return counts,total,damaged

tr,tot_tr,dam_tr=annotation_stats(TRAIN_LABELS); va,tot_va,dam_va=annotation_stats(VAL_LABELS)
display(pd.DataFrame({'Class':[CLASS_NAMES[i] for i in CLASS_NAMES], 'Train':[tr[i] for i in CLASS_NAMES], 'Validation':[va[i] for i in CLASS_NAMES]}))
print('Train objects:',tot_tr,'| damaged images:',dam_tr)
print('Validation objects:',tot_va,'| damaged images:',dam_va)
```

## 3. Digital Image Processing

```python
def find_image(folder,stem):
    for ext in ['.jpg','.jpeg','.png','.JPG','.JPEG','.PNG']:
        p=folder/f'{stem}{ext}'
        if p.exists(): return p

sample_label=next(p for p in sorted(TRAIN_LABELS.glob('*.txt')) if read_labels(p))
sample_image=find_image(TRAIN_IMAGES,sample_label.stem)
img=cv2.cvtColor(cv2.imread(str(sample_image)),cv2.COLOR_BGR2RGB)
resize=cv2.resize(img,(IMG_SIZE,IMG_SIZE)); gray=cv2.cvtColor(resize,cv2.COLOR_RGB2GRAY)
gaussian=cv2.GaussianBlur(gray,(5,5),0)
clahe=cv2.createCLAHE(2.0,(8,8)); enhanced=clahe.apply(gray)
edges=cv2.Canny(enhanced,50,150)
_,th=cv2.threshold(enhanced,0,255,cv2.THRESH_BINARY+cv2.THRESH_OTSU)
kernel=np.ones((3,3),np.uint8); morph=cv2.morphologyEx(th,cv2.MORPH_OPEN,kernel)

items=[(img,'Original'),(resize,'Resize'),(gray,'Grayscale'),(gaussian,'Gaussian Blur'),(enhanced,'Contrast Enhancement'),(edges,'Canny'),(th,'Threshold'),(morph,'Morphology')]
fig,ax=plt.subplots(2,4,figsize=(16,8))
for a,(im,title) in zip(ax.ravel(),items): a.imshow(im,cmap='gray' if im.ndim==2 else None); a.set_title(title); a.axis('off')
plt.tight_layout(); plt.show()
```

## 4. YOLO annotation / bounding-box localization

```python
def xyxy(xc,yc,bw,bh,w,h):
    x1=max(0,int((xc-bw/2)*w)); y1=max(0,int((yc-bh/2)*h)); x2=min(w-1,int((xc+bw/2)*w)); y2=min(h-1,int((yc+bh/2)*h)); return x1,y1,x2,y2

vis=img.copy(); h,w=vis.shape[:2]
for cls,xc,yc,bw,bh in read_labels(sample_label):
    x1,y1,x2,y2=xyxy(xc,yc,bw,bh,w,h); cv2.rectangle(vis,(x1,y1),(x2,y2),(255,0,0),3); cv2.putText(vis,CLASS_NAMES[cls],(x1,max(25,y1-8)),cv2.FONT_HERSHEY_SIMPLEX,.7,(255,0,0),2)
plt.figure(figsize=(13,8)); plt.imshow(vis); plt.title('YOLO Bounding Boxes / Damage Localization'); plt.axis('off'); plt.show()
```

## 5. Create practical CNN crop dataset

```python
if CROP_DIR.exists(): shutil.rmtree(CROP_DIR)
for split in ['train','val']:
    for folder in CLASS_FOLDERS.values(): (CROP_DIR/split/folder).mkdir(parents=True,exist_ok=True)

def make_crops(image_dir,label_dir,out_dir,limits=None):
    counts={i:0 for i in CLASS_NAMES}; label_files=list(label_dir.glob('*.txt')); random.Random(SEED).shuffle(label_files)
    for lp in label_files:
        rows=read_labels(lp)
        if not rows: continue
        ip=find_image(image_dir,lp.stem)
        if ip is None: continue
        im=cv2.imread(str(ip));
        if im is None: continue
        h,w=im.shape[:2]
        for cls,xc,yc,bw,bh in rows:
            if limits and counts[cls]>=limits.get(cls,0): continue
            x1,y1,x2,y2=xyxy(xc,yc,bw,bh,w,h)
            if x2<=x1 or y2<=y1: continue
            crop=cv2.resize(im[y1:y2,x1:x2],(IMG_SIZE,IMG_SIZE))
            cv2.imwrite(str(out_dir/CLASS_FOLDERS[cls]/f'{lp.stem}_{counts[cls]:04d}.jpg'),crop)
            counts[cls]+=1
    return counts

train_counts=make_crops(TRAIN_IMAGES,TRAIN_LABELS,CROP_DIR/'train',{0:250,1:60,2:250,3:250})
val_counts=make_crops(VAL_IMAGES,VAL_LABELS,CROP_DIR/'val')
print('Train:',train_counts); print('Validation:',val_counts)
```

## 6. CNN datasets

```python
train_ds=tf.keras.utils.image_dataset_from_directory(CROP_DIR/'train',image_size=(IMG_SIZE,IMG_SIZE),batch_size=BATCH_SIZE,shuffle=True,seed=SEED)
val_ds=tf.keras.utils.image_dataset_from_directory(CROP_DIR/'val',image_size=(IMG_SIZE,IMG_SIZE),batch_size=BATCH_SIZE,shuffle=False)
CNN_CLASSES=train_ds.class_names
train_ds=train_ds.prefetch(tf.data.AUTOTUNE); val_ds=val_ds.prefetch(tf.data.AUTOTUNE)
print(CNN_CLASSES)

plt.figure(figsize=(10,8))
for i,folder in enumerate(CNN_CLASSES):
    f=next((CROP_DIR/'train'/folder).glob('*.jpg')); plt.subplot(2,2,i+1); plt.imshow(Image.open(f)); plt.title(CLASS_NAMES[i]); plt.axis('off')
plt.tight_layout(); plt.show()
```

## 7. CNN class weights

```python
y_train=[]
for i,folder in enumerate(CNN_CLASSES): y_train += [i]*len(list((CROP_DIR/'train'/folder).glob('*.jpg')))
classes=np.unique(y_train); weights=compute_class_weight('balanced',classes=classes,y=y_train); class_weights={int(k):float(v) for k,v in zip(classes,weights)}
print(class_weights)
```

## 8. Custom CNN

```python
cnn=keras.Sequential([layers.Input((IMG_SIZE,IMG_SIZE,3)),layers.Rescaling(1/255.),layers.Conv2D(32,3,activation='relu'),layers.MaxPooling2D(),layers.Conv2D(64,3,activation='relu'),layers.MaxPooling2D(),layers.Conv2D(128,3,activation='relu'),layers.MaxPooling2D(),layers.Flatten(),layers.Dense(128,activation='relu'),layers.Dropout(.5),layers.Dense(4,activation='softmax')])
cnn.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
cnn.summary()
```

```python
cnn_history=cnn.fit(train_ds,validation_data=val_ds,epochs=15,class_weight=class_weights,callbacks=[keras.callbacks.EarlyStopping(monitor='val_loss',patience=3,restore_best_weights=True)])
```

```python
plt.figure(figsize=(12,5)); plt.subplot(1,2,1); plt.plot(cnn_history.history['accuracy'],label='Train'); plt.plot(cnn_history.history['val_accuracy'],label='Validation'); plt.title('CNN Accuracy'); plt.legend(); plt.subplot(1,2,2); plt.plot(cnn_history.history['loss'],label='Train'); plt.plot(cnn_history.history['val_loss'],label='Validation'); plt.title('CNN Loss'); plt.legend(); plt.show()
```

```python
cnn_loss,cnn_acc=cnn.evaluate(val_ds,verbose=0); y_true=np.concatenate([y.numpy() for _,y in val_ds]); y_prob=cnn.predict(val_ds,verbose=0); y_pred=np.argmax(y_prob,1)
print(f'CNN validation accuracy: {cnn_acc:.4f}')
print(classification_report(y_true,y_pred,target_names=CNN_CLASSES,digits=4,zero_division=0))
plt.figure(figsize=(7,6)); sns.heatmap(confusion_matrix(y_true,y_pred),annot=True,fmt='d',xticklabels=CNN_CLASSES,yticklabels=CNN_CLASSES); plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title('CNN Confusion Matrix'); plt.show()
```

## 9. VGG16 Transfer Learning

```python
base=VGG16(weights='imagenet',include_top=False,input_shape=(IMG_SIZE,IMG_SIZE,3)); base.trainable=False
vgg=keras.Sequential([layers.Input((IMG_SIZE,IMG_SIZE,3)),layers.Lambda(preprocess_input),base,layers.GlobalAveragePooling2D(),layers.Dense(128,activation='relu'),layers.Dropout(.5),layers.Dense(4,activation='softmax')])
vgg.compile(optimizer=keras.optimizers.Adam(1e-4),loss='sparse_categorical_crossentropy',metrics=['accuracy']); vgg.summary()
```

```python
vgg_history=vgg.fit(train_ds,validation_data=val_ds,epochs=8,class_weight=class_weights,callbacks=[keras.callbacks.EarlyStopping(monitor='val_loss',patience=2,restore_best_weights=True)])
```

```python
vgg_loss,vgg_acc=vgg.evaluate(val_ds,verbose=0); vgg_pred=np.argmax(vgg.predict(val_ds,verbose=0),1)
print(f'VGG16 validation accuracy: {vgg_acc:.4f}')
print(classification_report(y_true,vgg_pred,target_names=CNN_CLASSES,digits=4,zero_division=0))
display(pd.DataFrame({'Model':['Custom CNN','VGG16 Transfer Learning'],'Validation Accuracy':[cnn_acc,vgg_acc]}))
```

## 10. YOLO data.yaml

```python
YOLO_DIR.mkdir(parents=True,exist_ok=True)
YOLO_YAML=YOLO_DIR/'roadguard_data.yaml'
data={'path':str(DATASET_DIR.resolve()),'train':'train/images','val':'val/images','test':'test/images','names':{0:'D00_Longitudinal_Crack',1:'D10_Transverse_Crack',2:'D20_Alligator_Crack',3:'D40_Pothole'}}
with open(YOLO_YAML,'w') as f: yaml.safe_dump(data,f,sort_keys=False)
print(YOLO_YAML.read_text())
```

## 11. YOLO object detection training

```python
# Lightweight model for practical termwork runtime.
yolo_model=YOLO('yolov8n.pt')
yolo_model.train(data=str(YOLO_YAML),epochs=10,imgsz=640,batch=8,patience=3,project=str(YOLO_DIR),name='roadguard_yolov8n',exist_ok=True,seed=SEED,plots=True)
```

```python
metrics=yolo_model.val(data=str(YOLO_YAML),split='val',imgsz=640)
print('YOLO validation completed.')
```

## 12. Final RoadGuard AI inference

```python
test_files=[p for p in TEST_IMAGES.iterdir() if p.suffix.lower() in {'.jpg','.jpeg','.png'}]
final_image=random.choice(test_files)
results=yolo_model.predict(source=str(final_image),conf=.25,imgsz=640,save=True,project=str(YOLO_DIR),name='final_prediction',exist_ok=True)
annotated=results[0].plot()
plt.figure(figsize=(14,9)); plt.imshow(cv2.cvtColor(annotated,cv2.COLOR_BGR2RGB)); plt.title(f'RoadGuard AI — {final_image.name}'); plt.axis('off'); plt.show()
```

```python
r=results[0]
if r.boxes is None or len(r.boxes)==0: print('No road damage detected.')
else:
    rows=[]
    for cls,conf,box in zip(r.boxes.cls.cpu().numpy(),r.boxes.conf.cpu().numpy(),r.boxes.xyxy.cpu().numpy()):
        x1,y1,x2,y2=box.astype(int); rows.append({'Class':CLASS_NAMES[int(cls)],'Confidence':round(float(conf),4),'X1':x1,'Y1':y1,'X2':x2,'Y2':y2})
    display(pd.DataFrame(rows))
```
